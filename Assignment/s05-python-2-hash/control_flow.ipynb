## 1. if / elif / else
Conditions are checked top to bottom; the first true branch runs.

```python
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "F"

for m in (95, 80, 60, 30):
    print(m, grade(m))
```

Truthiness and the conditional expression:

```python
items = []
if not items:
    print("empty list is falsy")
age = 20
status = "adult" if age >= 18 else "minor"
print(status)
```

## 2. for loop
Iterates over any iterable.

```python
for ch in "abc":
    print(ch, end=" ")
print()
for i in range(1, 10, 3):  # start, stop (exclusive), step
    print(i, end=" ")
print()
for i, fruit in enumerate(["apple", "banana"], start=1):
    print(i, fruit)
for k, v in {"a": 1, "b": 2}.items():
    print(k, v)
```

## 3. while loop
Runs while the condition stays true, so something inside must change it.

```python
n, steps = 27, 0
while n != 1:  # Collatz sequence
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print("steps:", steps)
```

## 4. break, continue, and loop-else

```python
for n in range(2, 20):
    if n % 2 == 0 and n != 2:
        continue          # skip even numbers
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break         # not prime
    else:                 # runs only if the loop did NOT break
        print(n, "is prime")
```

## 5. Nested loops and while True

```python
for i in range(1, 4):
    print(" ".join(f"{i * j:2}" for j in range(1, 4)))

count = 0
while True:
    count += 1
    if count == 3:
        break
print("stopped at", count)
```

## Exercise: FizzBuzz
For 1..15 print Fizz for multiples of 3, Buzz for 5, FizzBuzz for both.

```python
for i in range(1, 16):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```
