## 1. `if` / `elif` / `else`

**What it is:** a way to run different code depending on whether a condition is `True` or `False`.

**Syntax:**

```python
if condition_1:
    ...  # runs if condition_1 is True
elif condition_2:
    ...  # runs if condition_1 is False and condition_2 is True
else:
    ...  # runs if every condition above is False
```

**Behind the scenes:** Python checks the conditions from top to bottom, runs **only the first** block whose condition is `True`, then jumps past the whole statement. Both `elif` and `else` are optional.

````python
def grade(marks):
    """Return the letter grade for a mark out of 100."""
    if marks >= 90:
        return "A"
    elif marks >= 75:  # only checked when marks < 90
        return "B"
    elif marks >= 50:  # only checked when marks < 75
        return "C"
    else:              # every check above was False
        return "F"

for m in (95, 80, 60, 30):
    print(m, grade(m))
````

**Order matters.** With the checks written the other way round, `marks >= 50` catches 95 first, and everyone above 50 gets a C:

````python
def wrong_grade(marks):
    if marks >= 50:  # 95 >= 50 is True, so this runs and the rest is skipped
        return "C"
    elif marks >= 75:
        return "B"
    elif marks >= 90:
        return "A"
    return "F"

print(wrong_grade(95))  # C, not A
````

**Combining conditions** with `and`, `or` and `not`. For example, a leap year is divisible by 4, except for century years, which must also be divisible by 400.

````python
def is_leap_year(year):
    """Return True if year is a leap year."""
    return (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

for year in (2024, 2026, 1900, 2000):
    print(year, is_leap_year(year))
````

**Truthiness and the conditional expression.** Empty or zero values (`0`, `""`, `[]`, `{}`, `None`) count as `False` in a condition; everything else counts as `True`. A conditional expression picks between two values on one line.

````python
cart = []
if not cart:  # an empty list is falsy, so not cart is True
    print("Your cart is empty")

age = 20
status = "adult" if age >= 18 else "minor"  # value_if_true if condition else value_if_false
print(status)
````

## 2. `for` loops

**What it is:** runs a block once for **each item** in an iterable, such as a list, string, `range`, or dictionary.

**Syntax:**

```python
for item in iterable:
    ...  # item holds the current element on each pass
```

**Behind the scenes:** Python asks the iterable for its items one at a time. On each pass, the next item is assigned to the loop variable and the block runs. When no items are left, the loop ends.

````python
for ch in "abc":        # a string gives its characters
    print(ch, end=" ")  # end=" " prints a space instead of starting a new line
print()

for i in range(1, 10, 3):  # range(start, stop, step); stop is not included -> 1, 4, 7
    print(i, end=" ")
print()

for i, fruit in enumerate(["apple", "banana"], start=1):  # enumerate() adds a counter
    print(i, fruit)

for k, v in {"a": 1, "b": 2}.items():  # .items() gives (key, value) pairs
    print(k, v)
````

**The accumulator pattern:** create an empty result before the loop, then update it on each pass. This is how you compute totals, counts, maximums and new lists.

````python
marks = [72, 95, 58, 81, 64]

total = 0
highest = marks[0]
passed = []

for m in marks:
    total += m            # running total
    if m > highest:       # keep the biggest value seen so far
        highest = m
    if m >= 60:
        passed.append(m)  # build a new list

print("total:", total)
print("average:", total / len(marks))
print("highest:", highest)
print("passed:", passed)
````

## 3. `while` loops

**What it is:** repeats a block **as long as** its condition is `True`. Use it when you don't know in advance how many repetitions you need.

**Syntax:**

```python
while condition:
    ...  # something in here must eventually make condition False
```

**Behind the scenes:** before every pass, Python checks the condition. If it's `True`, the block runs and Python checks again. If it's `False`, the loop ends. If nothing ever changes the condition, the loop runs forever; in Jupyter, press the stop button to interrupt it.

````python
# Collatz sequence: halve n if it's even, otherwise do 3n + 1. How many steps until n is 1?
n, steps = 27, 0
while n != 1:
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print("steps:", steps)  # there's no way to know the count in advance, so a while loop fits
````

````python
# Take a number apart digit by digit: % 10 gives the last digit, // 10 removes it
number = 4096
digit_sum = 0
reversed_number = 0

while number > 0:
    last_digit = number % 10
    digit_sum += last_digit
    reversed_number = reversed_number * 10 + last_digit
    number //= 10  # same as number = number // 10

print("digit sum:", digit_sum)       # 4 + 0 + 9 + 6 = 19
print("reversed:", reversed_number)  # 6904
````

**Input validation** is a common use of `while`: keep asking until the user types something valid. `input()` can't be answered during **Run All**, so a list of answers stands in for what a user would type. [input_demo.py](input_demo.py) shows the real `input()`.

````python
typed_answers = ["abc", "-5", "42"]  # what the user "types", one answer per attempt

attempt = 0
while True:
    answer = typed_answers[attempt]  # in a real script: answer = input("Enter your age: ")
    attempt += 1
    if answer.isdigit():             # only digits, so it's a whole number of 0 or more
        age = int(answer)
        break                        # valid: stop asking
    print(f"'{answer}' is not a valid age, try again")

print("age:", age, "after", attempt, "attempts")
````

## 4. `break`, `continue`, and the loop `else`

| Statement | What it does |
|---|---|
| `break` | Leaves the loop immediately |
| `continue` | Skips the rest of this pass and moves on to the next item |
| `else` on a loop | Runs once after the loop finishes **without** hitting `break` |

Read a loop's `else` as **"no break"**. It's ideal for searches, where the `else` block handles the "not found" case.

````python
# Primes below 20: skip even numbers, and stop testing a number at its first divisor
for n in range(2, 20):
    if n % 2 == 0 and n != 2:
        continue                           # even numbers (except 2) can't be prime: skip
    for d in range(2, int(n ** 0.5) + 1):  # divisors only need checking up to the square root
        if n % d == 0:
            break                          # found a divisor, so n is not prime
    else:                                  # the inner loop never hit break
        print(n, "is prime")
````

````python
def find_student(names, target):
    """Print where target is in names, or that it's missing."""
    for position, name in enumerate(names):
        if name == target:
            print(f"found {target} at position {position}")
            break
    else:  # runs only if the loop ended without break
        print(f"{target} is not in the list")

students = ["asha", "ravi", "monal"]
find_student(students, "ravi")
find_student(students, "alex")
````

## 5. Nested loops

A loop inside another loop: for **each** pass of the outer loop, the inner loop runs all the way through. With 3 outer passes and 5 inner passes, the inner block runs 3 × 5 = 15 times.

````python
# Multiplication table: rows come from the outer loop, columns from the inner loop
for i in range(1, 4):
    for j in range(1, 6):
        print(f"{i * j:3}", end="")  # :3 pads each number to width 3, so the columns line up
    print()                          # end the row
````

````python
# A star pyramid: row r has 2r - 1 stars, centred in a width of 9
rows = 5
for r in range(1, rows + 1):
    stars = ""
    for _ in range(2 * r - 1):  # 1, 3, 5, 7, 9 ("*" * (2 * r - 1) does this in one step)
        stars += "*"
    print(f"{stars:^9}")
````

## 6. Function basics

**What it is:** a **function** is a named block of code that does one job. You define it once and call it whenever you need it, instead of copying the same lines around.

**Syntax:**

```python
def function_name(parameter1, parameter2):
    """Docstring: one line saying what the function does."""
    ...
    return result
```

**Returns:** whatever follows `return`. A function without a `return` gives back `None`.

**Behind the scenes:** `def` only *creates* the function and stores it under its name; the body doesn't run yet. A **call** like `function_name(5, 3)` assigns the arguments to the parameters, runs the body, and hands back the returned value in place of the call. Variables created inside the body are **local**: they disappear when the function returns.

### 6.1 Defining and calling

````python
def greet():
    """Print a greeting."""
    print("Welcome to the FDE bootcamp!")

greet()  # call it once...
greet()  # ...and again, without copying any code
````

### 6.2 Parameters and `return`

**Parameters** are the names in the definition; **arguments** are the values passed in the call. `return` hands a value back to the caller, which can store it, print it, or use it in a bigger expression.

````python
def rectangle_area(length, width):  # length and width are parameters
    """Return the area of a rectangle."""
    return length * width

area = rectangle_area(5, 3)  # 5 and 3 are arguments
print(area)
print(rectangle_area(2, 4) + rectangle_area(1, 1))  # returned values can be used in expressions
````

### 6.3 `print` vs `return`

`print` only shows a value on the screen. `return` gives the value back, so the rest of the program can use it.

````python
def add_and_print(a, b):
    print(a + b)  # shows the answer, but gives back nothing (None)

def add_and_return(a, b):
    return a + b  # gives the answer back to the caller

result_1 = add_and_print(2, 3)   # prints 5
result_2 = add_and_return(2, 3)  # prints nothing

print(result_1)  # None: there's nothing to reuse
print(result_2)  # 5
````

### 6.4 Default values and keyword arguments

A **default value** makes a parameter optional. A **keyword argument** names the parameter in the call, so the order doesn't matter.

````python
def power(base, exponent=2):  # exponent defaults to 2
    """Return base raised to exponent."""
    return base ** exponent

print(power(5))                   # 25: exponent uses its default
print(power(2, 10))               # 1024
print(power(exponent=3, base=2))  # 8: keyword arguments, in any order
````

### 6.5 Returning several values, and returning early

`return a, b, c` returns a tuple, which the caller can unpack into separate variables. A `return` anywhere in the body ends the function immediately.

````python
def min_max_average(numbers):
    """Return the smallest, largest and average of a list of numbers."""
    if not numbers:  # early return: there's nothing to compute for an empty list
        return None
    return min(numbers), max(numbers), sum(numbers) / len(numbers)  # a tuple of 3 values

low, high, avg = min_max_average([72, 95, 58, 81, 64])  # unpack the tuple
print(low, high, avg)
print(min_max_average([]))
````

### 6.6 Functions with loops and conditions

Functions and control flow work together: a function can contain loops and `if` statements, and one function can call another.

````python
def factorial(n):
    """Return n! = 1 * 2 * ... * n."""
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result


def is_prime(n):
    """Return True if n is a prime number."""
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False  # return inside a loop ends the whole function at once
    return True


def primes_up_to(limit):
    """Return a list of the primes from 2 to limit."""
    primes = []
    for n in range(2, limit + 1):
        if is_prime(n):  # one function calling another
            primes.append(n)
    return primes


print(factorial(5))  # 120
print(is_prime(29), is_prime(30))
print(primes_up_to(30))
````

### 6.7 Local variables

A variable created inside a function exists only while that function runs.

````python
def make_total(values):
    running_total = 0  # local variable: exists only while make_total() runs
    for v in values:
        running_total += v
    return running_total

print(make_total([1, 2, 3]))

try:
    print(running_total)  # not defined out here
except NameError as error:
    print("NameError:", error)
````

**1. FizzBuzz.** For 1 to 15, print `Fizz` for multiples of 3, `Buzz` for multiples of 5, `FizzBuzz` for multiples of both, and the number otherwise. Write a function that returns the right word for one number.

````python
# Solution 1: FizzBuzz
def fizzbuzz(n):
    """Return the FizzBuzz word for n."""
    if n % 15 == 0:  # check "both" first: 15 is also a multiple of 3 and of 5
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    return str(n)

for i in range(1, 16):
    print(fizzbuzz(i), end=" ")
print()
````

**2. Simple calculator.** Write `calculate(a, operator, b)` that supports `+`, `-`, `*` and `/`, and returns an error message for division by zero or an unknown operator.

````python
# Solution 2: calculator with if / elif / else
def calculate(a, operator, b):
    """Apply operator to a and b, or return an error message."""
    if operator == "+":
        return a + b
    elif operator == "-":
        return a - b
    elif operator == "*":
        return a * b
    elif operator == "/":
        if b == 0:  # a nested if guards against dividing by zero
            return "Error: cannot divide by zero"
        return a / b
    else:
        return f"Error: unknown operator '{operator}'"

print(calculate(8, "+", 2))
print(calculate(8, "/", 2))
print(calculate(8, "/", 0))
print(calculate(8, "^", 2))
````

**3. Count vowels.** Write a function that counts the vowels in a string, ignoring upper and lower case.

````python
# Solution 3: count vowels with a for loop and the in operator
def count_vowels(text):
    """Return how many vowels (a, e, i, o, u) text contains."""
    count = 0
    for char in text.lower():  # lower() so "E" and "e" both count
        if char in "aeiou":
            count += 1
    return count

print(count_vowels("Forward Deployed Engineer"))
````

**4. Palindrome number.** Check whether a number reads the same backwards (like 12321), using a `while` loop and no strings.

````python
# Solution 4: palindrome check with digit arithmetic
def is_palindrome_number(n):
    """Return True if the digits of n read the same in both directions."""
    original, reversed_n = n, 0
    while n > 0:
        reversed_n = reversed_n * 10 + n % 10  # append n's last digit
        n //= 10                               # drop n's last digit
    return original == reversed_n

for number in (12321, 4096, 7):
    print(number, is_palindrome_number(number))
````

**5. Number guessing game.** The secret number is 37. Go through a list of guesses (standing in for `input()`), print "too low" or "too high" for each one, and stop at the right guess. Report the number of attempts, or that the player ran out of guesses.

````python
# Solution 5: guessing game with while, if / elif / else, break and the loop else
def play(secret, guesses, max_attempts=5):
    """Print feedback for each guess until it's right or the attempts run out."""
    attempt = 0
    while attempt < max_attempts and attempt < len(guesses):
        guess = guesses[attempt]  # in a real game: guess = int(input("Your guess: "))
        attempt += 1
        if guess < secret:
            print(f"{guess}: too low")
        elif guess > secret:
            print(f"{guess}: too high")
        else:
            print(f"{guess}: correct, in {attempt} attempts!")
            break
    else:  # the loop ended without break: the number was never guessed
        print(f"Out of guesses. The number was {secret}.")

play(37, [50, 25, 37])
print()
play(37, [10, 90, 20, 80, 30])
````

**6. Class report.** Put it all together: for a dict of student marks, print each student's grade using `grade()` from section 1, then the class average and the top scorer.

````python
# Solution 6: functions, loops and conditions together
def class_report(marks_by_student):
    """Print each student's grade, then the class average and the top scorer."""
    top_student, top_mark, total = None, -1, 0
    for student, marks in marks_by_student.items():
        print(f"{student:<6} {marks:>3}  {grade(marks)}")
        total += marks
        if marks > top_mark:
            top_student, top_mark = student, marks
    print(f"average: {total / len(marks_by_student):.1f}")
    print(f"top scorer: {top_student} ({top_mark})")

class_report({"asha": 92, "ravi": 48, "monal": 77, "alex": 63})
````

## Summary

- `if` / `elif` / `else` runs only the **first** block whose condition is `True`, so put the strictest check first.
- Empty and zero values are falsy; `x if condition else y` picks a value on one line.
- `for` loops over the items of an iterable; the accumulator pattern builds totals, maximums and lists.
- `while` repeats until its condition becomes `False`; something in the loop must change it, or `break` must end it.
- `break` leaves a loop, `continue` skips to the next pass, and a loop's `else` runs only when there was no `break`.
- A nested loop runs completely for every pass of the outer loop.
- `def` creates a function; calling it runs the body. `return` gives a value back (`None` if there's no `return`), and `return a, b` returns a tuple.
- Parameters can have default values, and arguments can be passed by keyword. Variables inside a function are local to it.
