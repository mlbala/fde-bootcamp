## 1. Conditional Statements

An `if` statement runs a block of code only when its condition is `True`.

```python
if condition:
    # runs only when condition is True
    statement
```

- The `if` line ends with a colon `:`.
- The block underneath is **indented** (4 spaces by convention). Indentation is how Python knows which lines belong to the `if`; Python does not use curly braces.

### 1.1 if

````python
first_class = True

# The condition is True, so the indented line runs
if first_class:
    print("Complete the backlog")
````

````python
class_no = 5

# Combine conditions with and: both must be True
if (class_no > 3) and (class_no <= 5):
    print("Welcome back! We are refreshing Python")

# The same check written as a chained comparison
if 3 < class_no <= 5:
    print("Chained comparisons work too")
````

````python
# The simplest possible if: the condition is always True
if True:
    print("Hello")
````

### 1.2 if-else

`else` runs when the `if` condition is `False`, so exactly one of the two blocks runs.

````python
class_no = 5  # try 2

# if / else: exactly one of the two blocks runs
if (class_no > 3) and (class_no <= 5):
    print("Welcome back! We are refreshing Python")
else:
    print("Please complete the backlog")

# A separate if statement: it is checked on its own, whatever happened above
if class_no == 5:
    print("Welcome back! You are on track")
````

### 1.3 if-elif-else

`elif` ("else if") adds more conditions. Python checks them from top to bottom and runs only the **first** block whose condition is `True`; the rest are skipped. `else` catches everything that did not match.

Because only the first match runs, order matters: put the strictest condition first.

````python
marks = 75  # try 95, 55, or 30

if marks >= 90:
    print("Grade A")
elif marks >= 70:  # checked only when marks < 90
    print("Grade B")
elif marks >= 50:  # checked only when marks < 70
    print("Grade C")
else:              # runs when none of the conditions above is True
    print("Not good")

# This line is not indented, so it is outside the if block and always runs
print("Control flow completed")
````

### 1.4 Nested if

An `if` can sit inside another `if`. The inner condition is checked only when the outer condition is `True`.

````python
age = 35        # try 15
has_id = True   # try False

if age >= 18:
    # Only reached when age >= 18
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Underage, not allowed")
````

### 1.5 Truthiness

A condition does not have to be a `bool`. When a value is used in an `if` or `while`, Python treats it as either **truthy** (acts like `True`) or **falsy** (acts like `False`).

| Falsy (treated as `False`) | Truthy (treated as `True`) |
|---|---|
| `False`, `None` | `True` |
| `0`, `0.0` | any non-zero number, such as `1`, `-5`, `1.0` |
| the empty string `""` | any non-empty string, such as `"a"` or `" "` |
| empty collections: `[]`, `()`, `{}`, `set()` | any non-empty collection, such as `[0]` |

`bool(value)` shows how Python will treat a value.

````python
# bool() shows whether a value is truthy or falsy
print(bool(0), bool(""), bool([]), bool(None))    # all falsy  -> False
print(bool(5), bool("hi"), bool([0]), bool(" "))  # all truthy -> True
````

````python
name = ""  # try "Monal"

# An empty string is falsy, so the else block runs
if name:
    print(f"Hello, {name}")
else:
    print("Name is empty")
````

### 1.6 Conditional expression

A **conditional expression** (also called a ternary operator) is a one-line `if-else` that produces a value:

```python
value_if_true if condition else value_if_false
```

Use it for a simple choice between two values. For anything longer, a regular `if-else` is easier to read.

````python
age = 20

# Regular if-else: four lines
if age >= 18:
    status = "Adult"
else:
    status = "Minor"
print(status)
````

````python
# Conditional expression: the same logic in one line
status = "Adult" if age >= 18 else "Minor"
print(status)
````

## 2. Loops

A **loop** repeats a block of code.

- A `for` loop runs once for each item in a sequence (a list, string, range, dictionary, and so on). Use it when you know what you are looping over.
- A `while` loop runs as long as a condition stays `True`. Use it when you do not know in advance how many repetitions you need.

### 2.1 range()

`range(start, stop, step)` produces a sequence of integers:

- `start`: the first number (default `0`)
- `stop`: the sequence ends **before** this number, so `stop` itself is never included
- `step`: the gap between numbers (default `1`; a negative step counts down)

`range` is lazy: it produces the numbers one at a time instead of storing them all in memory. Wrap it in `list()` to see every number.

````python
# On its own, range() shows only the range object, not the numbers
range(1, 11)
````

````python
print(list(range(1, 11)))      # 1 up to (not including) 11
print(list(range(5)))          # only stop given: starts at 0 -> 0 to 4
print(list(range(1, 11, 2)))   # step of 2 -> 1, 3, 5, 7, 9
print(list(range(10, 0, -3)))  # negative step counts down -> 10, 7, 4, 1
````

### 2.2 for loop

```python
for variable in sequence:
    # runs once per item; variable holds the current item
    statement
```

````python
# The body runs once for each number in range(1, 4), so 3 times
for i in range(1, 4):
    print("FDE")
````

````python
# The loop variable i holds the current number on each pass
for i in range(1, 4):
    print(i)
    print("FDE")
````

The loop variable can have any valid name. Be careful: a loop variable still exists after the loop ends, and it keeps its last value.

````python
# The loop variable here is number, but the body prints i
for number in range(1, 4):
    print(i)  # i is left over from the previous loop, so it is still 3
    print("FDE")

print(number)  # the loop variable keeps its last value after the loop -> 3
````

````python
# A step of 2 gives every other number
for i in range(1, 11, 2):
    print(i)  # 1, 3, 5, 7, 9
````

### 2.3 pass, continue, and break

| Statement | What it does |
|---|---|
| `pass` | Does nothing. It is a placeholder for where Python needs a statement but you have nothing to run yet |
| `continue` | Skips the rest of the current iteration and jumps to the next one |
| `break` | Exits the loop immediately; execution continues with the first line after the loop |

The three cells below run the same loop. Only the statement used for certain numbers changes, so compare their outputs.

````python
list_values = list(range(0, 11))

for i in list_values:
    if i in [3, 4, 5]:
        pass  # does nothing: the loop carries on to the next line as normal
    else:
        print(i)
    print("Iteration completed")  # still runs for 3, 4 and 5
````

````python
for i in list_values:
    if i in [3, 4, 5]:
        continue  # skip the rest of this iteration and go back to the for line
    else:
        print(i)
    print("Iteration completed")  # skipped for 3, 4 and 5
````

````python
for i in list_values:
    if i == 3:
        break  # leave the loop entirely and jump to the first line after it
    else:
        print(i)
    print("Iteration completed")

print("Loop completed.")  # runs right after the break
````

### 2.4 Nested loops

A loop can sit inside another loop. For every iteration of the outer loop, the inner loop runs from start to finish. `break` exits only the **innermost** loop it is in.

````python
# Problem statement:
# Iterate over the numbers 0 to 10.
# When the number is 7, print "Welcome to FDE bootcamp" 3 times.

for i in range(0, 11):           # outer loop
    print(i)
    if i == 7:
        for j in range(0, 300):  # inner loop
            if j == 3:
                break            # exits only the inner loop; the outer loop carries on
            print("Welcome to FDE bootcamp")
````

````python
# A simpler solution: range(3) repeats exactly 3 times, so no break is needed
for i in range(0, 11):
    print(i)
    if i == 7:
        for _ in range(3):  # _ is the convention for a loop variable you don't use
            print("Welcome to FDE bootcamp")
````

### 2.5 while loop

```python
while condition:
    # runs again and again while condition is True
    statement
```

Something inside the loop must eventually make the condition `False` or call `break`. Otherwise the loop never ends (an **infinite loop**). If that happens in Jupyter, press the stop button (■) or choose **Kernel → Interrupt**.

````python
# A while loop with a counter: it runs while i is less than 5
i = 0
while i < 5:
    print("hello", i)
    i = i + 1  # move i towards the stop condition (can also be written i += 1)
````

````python
# while True never becomes False on its own, so break is the only way out
i = 0
while True:
    print("hello")
    i = i + 1
    if i > 10:
        break  # stops after 11 prints (i = 0 to 10)
````

### 2.6 Looping over collections

A `for` loop works with any **iterable**: lists, tuples, strings, sets, and dictionaries.

````python
# Looping over a list gives each item
for i in [1, 2, 3, 4, 5]:
    print(i)
````

````python
# Looping over a string gives each character
for char in "FDE":
    print(char)
````

````python
students = {"101": "Ananda", "102": "Yuval", "103": "Anurag"}

# Looping over a dictionary gives its keys
for roll_no in students:
    print(roll_no)
````

**Unpacking** assigns the items of a tuple (or any sequence) to several variables in one line. The number of variables must match the number of items.

````python
a = (10, 2, 4)
print(a)  # one variable holds the whole tuple
````

````python
# Unpacking: each variable gets one item, in order
a1, b1, c1 = (10, 2, 4)
print(a1, b1, c1)
````

`dict.items()` gives `(key, value)` pairs, so you can unpack each pair straight into two loop variables.

````python
count = 0

# Each item is a (key, value) tuple, unpacked into roll_no and name
for roll_no, name in students.items():
    print(roll_no, name)
    count += 1  # same as count = count + 1

print(f"Loop executed {count} times")
````

`enumerate()` adds a counter to any iterable and gives `(index, item)` pairs. With `.items()`, each item is itself a pair, so every step has the shape `(index, (key, value))`. That is **two** values, not three, so unpacking into three flat variables fails.

````python
# enumerate gives 2 values per step: (index, (key, value))
# Asking for 3 flat variables raises a ValueError
try:
    for i, roll_no, name in enumerate(students.items()):
        print(i, roll_no, name)
except ValueError as error:
    print("ValueError:", error)
````

````python
# Fix: mirror the structure with parentheses -> index, (key, value)
# start=1 makes the counter begin at 1 instead of 0
for i, (roll_no, name) in enumerate(students.items(), start=1):
    print(i, roll_no, name)

print(f"Loop executed {i} times")
````

## 3. Functions

A **function** is a named, reusable block of code. Instead of repeating the same lines, you write them once inside a function and **call** the function whenever you need it.

```python
def function_name(parameters):
    # body: indented code that runs when the function is called
    return result
```

- `def` starts the definition, and the body is indented.
- **Parameters** are the names listed in the definition. **Arguments** are the actual values you pass in when you call the function.
- `return` sends a value back to the caller and ends the function. A function without `return` gives back `None`.
- Defining a function does not run it. The body runs only when you call it: `function_name(arguments)`.

````python
# Without a function: the calculation is written out directly
price_1 = 100
quantity_1 = 5

total_price_1 = price_1 * quantity_1
print(total_price_1)
````

````python
# The same steps wrapped in a function; defining it prints nothing
def calculate_total():
    price_1 = 100
    quantity_1 = 5
    total_price_1 = price_1 * quantity_1
    print(total_price_1)

# Calling the function runs its body
calculate_total()
````

The function above always uses the same price and quantity. **Parameters** make it work for any values, and `return` hands the result back so the caller can store it or use it again.

````python
def calculate_total(price, quantity):
    # Send the result back to whoever called the function
    return price * quantity

total = calculate_total(100, 5)  # store the returned value in a variable
print(total)                     # 500
print(calculate_total(250, 3))   # use the returned value directly -> 750
````

### 3.1 More ways to define functions

A function can take no parameters, return nothing, return several values at once, or stop early with `return`. A string in triple quotes on the first line of the body is a **docstring**: it documents what the function does, and tools such as `help()` display it.

````python
# No parameters and no return statement
def say_hello():
    print("Hello, FDE bootcamp!")

result = say_hello()  # the body runs and prints the greeting
print(result)         # a function without return gives back None
````

````python
def area_of_rectangle(length, width):
    """Return the area of a rectangle."""  # docstring: describes the function
    return length * width

print(area_of_rectangle(4, 5))    # 20
print(area_of_rectangle.__doc__)  # the docstring is stored on the function (help() shows it too)
````

````python
def min_max(numbers):
    """Return the smallest and the largest number in a list."""
    return min(numbers), max(numbers)  # several values are returned together as a tuple

result = min_max([4, 9, 1, 7])
print(result)  # (1, 9)

# Unpack the returned tuple straight into two variables
smallest, largest = min_max([4, 9, 1, 7])
print(smallest, largest)
````

````python
def check_number(n):
    """Describe whether n is positive, negative, or zero."""
    if n > 0:
        return "positive"  # return ends the function immediately
    if n < 0:
        return "negative"
    return "zero"          # reached only when both checks above are False

print(check_number(5))
print(check_number(-3))
print(check_number(0))
````

````python
def is_even(n):
    """Return True if n is even."""
    return n % 2 == 0  # a comparison already gives True or False

def count_even(numbers):
    """Count how many numbers in the list are even."""
    count = 0
    for n in numbers:
        if is_even(n):  # one function can call another
            count += 1
    return count

print(count_even([1, 2, 3, 4, 5, 6]))  # 3
````

````python
# A function is an object: it can be stored in a variable or passed to another function
def square(n):
    return n * n

my_function = square   # no parentheses: this refers to the function without calling it
print(my_function(6))  # 36

def apply_to_all(func, values):
    """Call func on every value and return the results in a list."""
    results = []
    for value in values:
        results.append(func(value))
    return results

print(apply_to_all(square, [1, 2, 3]))  # [1, 4, 9]
````

### 3.2 Positional arguments

Arguments are matched to parameters by **position**: the first argument goes to the first parameter, the second to the second, and so on.

The `add()` function below both **prints** and **returns**. The printed lines appear first. The returned value is shown afterwards as the cell's output, because the call is the last line of the cell.

````python
def add(a, b):
    print(a)      # print() only displays a value
    print(b)
    return a + b  # return sends the result back to the caller

# 10 goes to a, 2 goes to b
add(10, 2)
````

### 3.3 Keyword arguments

With **keyword arguments**, you name each parameter in the call (`name=value`). The order no longer matters, and the call is easier to read.

````python
# Matched by name, so the order does not matter
add(b=3, a=2)
````

````python
# Positional and keyword arguments can be mixed, but the positional ones must come first
add(2, b=3)

# add(a=2, 3)  # SyntaxError: positional argument follows keyword argument
````

### 3.4 Missing arguments

Every parameter without a default value is **required**. Calling the function with too few arguments raises a `TypeError`.

````python
# No arguments: both a and b are missing
try:
    add()
except TypeError as error:
    print("TypeError:", error)

# One argument: b is missing
try:
    add(1)
except TypeError as error:
    print("TypeError:", error)
````

### 3.5 Default arguments

A parameter can have a **default value** (`parameter=value`). If the caller leaves that argument out, the default is used.

Parameters with defaults must come **after** parameters without defaults: `def greet(name, message="Hi")` is valid, but `def greet(message="Hi", name)` is a `SyntaxError`.

````python
# Both parameters have defaults, so every argument is optional
def add(a=0, b=0):
    print(a)
    print(b)
    return a + b
````

````python
add(10, 10)  # both defaults are replaced -> 20
````

````python
# name is required; message is optional
def greet(name, message="great day!"):
    print(message, name)
````

````python
greet("saurabh", "great day!")  # both arguments given
greet("amit")                   # message left out, so the default is used
greet("amit", "Hi")             # the default is replaced with "Hi"
````

**Watch out for mutable defaults.** A default value is created **once**, when the function is defined, not every time the function is called. So a list used as a default is shared by every call. Use `None` as the default and create the list inside the function instead.

````python
# Bug: the default list is created once and shared by every call
def add_item_buggy(item, items=[]):
    items.append(item)
    return items

print(add_item_buggy("pen"))   # ['pen']
print(add_item_buggy("book"))  # ['pen', 'book']: 'pen' from the previous call is still there
````

````python
# Fix: use None as the default and create a new list inside the function
def add_item(item, items=None):
    if items is None:
        items = []  # a fresh list on every call
    items.append(item)
    return items

print(add_item("pen"))   # ['pen']
print(add_item("book"))  # ['book']
````

### 3.6 Variable positional arguments: *args

The `total()` function below has four parameters, so it accepts at most four values.

````python
def total(a=0, b=0, c=0, d=0):
    print(a + b + c + d)

total(10, 20, 30, 40)  # 100
total()                # 0: every parameter uses its default
total(10)              # 10: b, c and d use their defaults
````

````python
# Only 4 parameters, so a 5th value has nowhere to go
try:
    total(10, 20, 30, 40, 50)
except TypeError as error:
    print("TypeError:", error)
````

`*args` collects any **extra** positional arguments into a **tuple**. The name `args` is just a convention; the `*` is what does the work.

````python
def total(a=0, b=0, c=0, d=0, *args):
    print(args)           # the extra values, collected in a tuple
    print(a + b + c + d)  # only the first four values are added here

total(1, 2, 3, 4, 5, 6, 7, 8, 9)
# a = 1, b = 2, c = 3, d = 4
# args = (5, 6, 7, 8, 9)
````

````python
total(1, 2, 3, 4)  # no extra values, so args is an empty tuple ()
````

````python
# With only *args, the function accepts any number of values
def total(*args):
    print(args)
    return sum(args)  # sum() adds up every item in the tuple
````

````python
total(1)  # a single value still arrives as a tuple: (1,)
````

````python
total(1, 2, 3, 4, 5, 6, 7, 8, 19, 3232, 32)
````

````python
def average(*numbers):
    """Return the average of any number of values (0 when none are given)."""
    if not numbers:  # an empty tuple is falsy
        return 0
    return sum(numbers) / len(numbers)

print(average(70, 80, 90))  # 80.0
print(average())            # 0
````

A `*` in a function **call** does the opposite of `*args`: it spreads a list or tuple out into separate arguments.

````python
marks_list = [70, 80, 90]

# Same as total(70, 80, 90)
total(*marks_list)
````

### 3.7 Variable keyword arguments: **kwargs

`**kwargs` collects any **extra** keyword arguments into a **dictionary** of `name: value` pairs. As with `args`, the name `kwargs` is a convention; the `**` is what does the work.

````python
def show_user_info(name, phone_no, **kwargs):
    print(kwargs)        # the extra keyword arguments, collected in a dict
    print(type(kwargs))  # <class 'dict'>
    print(name, phone_no)
````

````python
# No extra keyword arguments, so kwargs is an empty dict
show_user_info(phone_no="+983232323", name="suhas")
````

````python
# email does not match any named parameter, so it lands in kwargs
show_user_info(phone_no="+983232323", name="suhas", email="suhas@example.com")
````

````python
def build_profile(name, **details):
    """Return a dict with the name plus any extra details passed in."""
    profile = {"name": name}
    for key, value in details.items():  # kwargs is a normal dict, so you can loop over it
        profile[key] = value
    return profile

print(build_profile("monal", city="Pune", role="FDE"))
````

A `**` in a function **call** spreads a dictionary out into keyword arguments.

````python
user = {"name": "suhas", "phone_no": "+983232323", "email": "suhas@example.com"}

# Same as show_user_info(name="suhas", phone_no="+983232323", email="suhas@example.com")
show_user_info(**user)
````

### 3.8 Combining *args and **kwargs

A function can accept both. The order in the definition is: regular parameters, then `*args`, then any keyword-only parameters, then `**kwargs`. A parameter placed after `*args` can only be passed by keyword, because `*args` takes every remaining positional value.

````python
def show_all(*args, **kwargs):
    print(args)    # positional values -> tuple
    print(kwargs)  # keyword values    -> dict

show_all(1, 2, 3, 4, 5, 6, 7, name="monal", phone_no="+8908029w")
````

````python
# Order: regular parameter, *args, keyword-only parameter, **kwargs
def place_order(customer, *items, discount=0, **notes):
    print("customer:", customer)
    print("items:", items)        # extra positional values -> tuple
    print("discount:", discount)  # comes after *items, so it can only be passed by keyword
    print("notes:", notes)        # extra keyword values -> dict

place_order("monal", "pen", "book", "bag", discount=10, gift_wrap=True)
````

### 3.9 Lambda functions

A **lambda** is a small, anonymous (unnamed) function written on one line:

```python
lambda parameters: expression
```

It returns the value of the expression automatically, with no `return` keyword. Lambdas are handy for short functions you pass to another function, such as the `key` argument of `sorted()`. For anything longer than one expression, use `def`.

````python
# def version
def square(n):
    return n * n

# lambda version: the same behaviour in one line
square_lambda = lambda n: n * n

print(square(5), square_lambda(5))  # 25 25
````

````python
student_marks = [("Ananda", 82), ("Yuval", 95), ("Anurag", 74)]

# key tells sorted() what to compare: here, the marks at index 1 of each tuple
by_marks = sorted(student_marks, key=lambda student: student[1])
print(by_marks)
````

## 4. Scope of a Variable

**Scope** is the part of a program where a variable can be seen and used.

- A **global** variable is created at the top level of a script or notebook. It can be read from anywhere, including inside functions.
- A **local** variable is created inside a function. It exists only while the function runs and cannot be seen from outside.

When Python looks up a name, it searches the scopes in this order, known as the **LEGB rule**:

| Order | Scope | Where the name was created |
|---|---|---|
| 1 | **L**ocal | inside the current function |
| 2 | **E**nclosing | inside an outer function, when one function is defined inside another |
| 3 | **G**lobal | at the top level of the script or notebook |
| 4 | **B**uilt-in | Python's own names, such as `print` and `len` |

### 4.1 Local variables

Any variable assigned inside a function, including its parameters, is **local** to that function.

````python
a = 10  # global variable

def do_something():
    # Assigning to a inside the function creates a NEW local variable a
    a = 0
    a = a + 1
    print(a)  # the local a -> 1

do_something()
print(a)  # the global a was never touched -> 10
````

````python
def make_greeting():
    message = "Hello from inside the function"  # local variable
    print(message)

make_greeting()

# message existed only while make_greeting() was running
try:
    print(message)
except NameError as error:
    print("NameError:", error)
````

````python
def add_bonus(score):
    bonus = 5              # local variable
    score = score + bonus  # parameters are local variables too
    return score

marks = 80
print(add_bonus(marks))  # 85
print(marks)             # 80: changing the parameter did not change the caller's variable
````

````python
def count_calls():
    calls = 0  # created fresh every time the function runs
    calls += 1
    print("calls:", calls)

count_calls()  # calls: 1
count_calls()  # calls: 1 again, because local variables don't survive between calls
````

````python
# The same name in two functions gives two separate local variables
def morning():
    greeting = "Good morning"  # local to morning()
    print(greeting)

def evening():
    greeting = "Good evening"  # a different variable, local to evening()
    print(greeting)

morning()
evening()
````

### 4.2 Global variables

To change a global variable from inside a function, declare it with the `global` keyword first.

````python
a = 10

def do_something():
    global a   # a now refers to the global variable
    a = a + 1  # so this updates the global a
    print(a)   # 11

do_something()
print(a)  # the global a was changed -> 11
````

Reading a global variable inside a function is allowed without `global`.

````python
a = 10

def do_something():
    # Reading a global variable inside a function is allowed
    print(a)

do_something()
print(a)
````

However, if a function **assigns** to a name anywhere in its body, Python treats that name as local for the **whole** function. Reading it before the assignment then fails with an `UnboundLocalError`.

````python
a = 10

def do_something():
    # a is assigned in this function, so Python treats a as local everywhere in it
    a = a + 1  # reads the local a before it has a value -> error
    print(a)

try:
    do_something()
except UnboundLocalError as error:
    print("UnboundLocalError:", error)
````

A mutable object such as a list can be **changed in place** from inside a function without `global`. Methods like `append()` modify the existing object; they do not assign a new value to the name.

````python
a = [1, 2, 3]

def do_something():
    # append() changes the existing list and does not reassign a, so no global is needed
    a.append(123)
    print(a)

do_something()
print(a)  # the global list was changed
````

````python
visit_count = 0  # global counter

def record_visit():
    global visit_count  # needed because the function assigns to visit_count
    visit_count += 1

record_visit()
record_visit()
record_visit()
print(visit_count)  # 3
````

````python
def set_course():
    global course_name  # course_name doesn't exist yet, so this creates it as a global
    course_name = "FDE Bootcamp"

set_course()
print(course_name)  # visible outside the function
````

### 4.3 Local (nested) functions and the enclosing scope

A function defined inside another function is a **local (nested) function**. Like a local variable, it exists only inside the outer function, so it cannot be called from outside. Nested functions are useful for small helpers that only one function needs.

````python
def print_report(scores):
    # A local helper function: it exists only inside print_report()
    def grade(score):
        return "Pass" if score >= 50 else "Fail"

    for name, score in scores.items():
        print(name, score, grade(score))

print_report({"Ananda": 82, "Yuval": 45, "Anurag": 67})

# grade() is local to print_report(), so it is not available out here
try:
    grade(70)
except NameError as error:
    print("NameError:", error)
````

````python
def outer():
    message = "set in outer"  # the enclosing variable for inner()

    def inner():
        print(message)  # inner can READ the enclosing variable without any keyword

    inner()

outer()
````

To **change** an enclosing variable from the inner function, declare it with `nonlocal`.

````python
def outer():
    count = 0  # local to outer, and the enclosing scope for inner

    def inner():
        nonlocal count  # refers to outer's count instead of creating a new local
        count += 1

    inner()
    inner()
    print(count)  # inner ran twice -> 2

outer()
````

An outer function can also **return** its inner function. The inner function remembers the enclosing variables even after the outer function has finished running. This is called a **closure**.

````python
def make_multiplier(factor):
    def multiply(n):
        return n * factor  # factor comes from the enclosing scope
    return multiply        # return the function itself, without calling it

double = make_multiplier(2)  # a function that remembers factor = 2
triple = make_multiplier(3)  # a function that remembers factor = 3

print(double(10))  # 20
print(triple(10))  # 30
````

### 4.4 Built-in scope and the full LEGB lookup

The built-in scope is the last place Python looks. If you create a variable with the same name as a built-in such as `sum`, `list` or `len`, your variable **shadows** (hides) the built-in.

````python
sum = 100  # shadows the built-in sum() function with a number

try:
    sum([1, 2, 3])  # sum is now an int, which cannot be called
except TypeError as error:
    print("TypeError:", error)

del sum  # delete our variable, so the name refers to the built-in again
print(sum([1, 2, 3]))  # 6
````

Here, the same name `x` exists at every level. Python uses the **nearest** one, searching Local → Enclosing → Global → Built-in.

````python
x = "global"

def outer():
    x = "enclosing"

    def inner():
        x = "local"
        print("inner sees:", x)  # L: inner's own x wins

    inner()
    print("outer sees:", x)      # outer's own x

outer()
print("top level sees:", x)      # G: the global x
print(len("LEGB"))               # B: len is found in the built-in scope
````

````python
x = "global"

def outer():
    def inner():
        print("inner sees:", x)  # no local or enclosing x, so Python falls back to the global one
    inner()

outer()
````

### 4.5 Blocks do not create a new scope

In Python, `if`, `for`, and `while` blocks do **not** create a new scope (unlike some other languages). A variable created inside them is still available after the block ends. Only functions (and classes and modules) create a new scope.

````python
if True:
    inside_if = "created inside the if block"

for k in range(3):
    inside_loop = k * 10

print(inside_if)    # still available after the if block
print(inside_loop)  # 20: the value from the last iteration
print(k)            # 2: the loop variable stays too
````

### 4.6 Good practice

Avoid `global` where you can. A function is easier to understand and test when it takes its inputs as parameters and sends its result back with `return`.

````python
# Preferred: pass the value in and return the new value, instead of using global
def increment(value):
    return value + 1

a = 10
a = increment(a)
print(a)  # 11
````

## Summary

- `if` / `elif` / `else` run the **first** block whose condition is `True`. Indentation decides which lines belong to a block.
- Empty or zero values (`0`, `""`, `[]`, `None`, ...) are **falsy**; everything else is **truthy**.
- `x if condition else y` is a one-line `if-else` that produces a value.
- `for` loops over the items of an iterable; `while` repeats as long as a condition is `True`.
- `range(start, stop, step)` never includes `stop`.
- `break` exits the loop, `continue` skips to the next iteration, and `pass` does nothing.
- Functions are defined with `def` and send results back with `return`. A function without `return` gives back `None`, and `return a, b` returns a tuple.
- Arguments can be positional, keyword, or default. `*args` collects extra values into a tuple and `**kwargs` into a dict. Never use a mutable default like `[]`; use `None` instead.
- `lambda parameters: expression` is a one-line anonymous function, handy as the `key` for `sorted()`.
- Local variables and nested functions exist only inside their function and are created fresh on every call.
- Names are looked up in **LEGB** order. Assigning inside a function creates a local variable unless you declare `global` or `nonlocal`; changing a list in place needs neither. `if` and `for` blocks do not create a new scope.
