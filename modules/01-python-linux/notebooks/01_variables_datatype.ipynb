## 1. Variables

A **variable** is a name that refers to a value stored in memory. You create one with the assignment operator `=`: the name goes on the left and the value on the right.

Python is **dynamically typed**, so you never declare a type. Python works out the type from the value, and the same name can later refer to a value of a different type.

```python
# Assignment: the name on the left now refers to the value on the right
a = 3
b = 4
```

`print()` displays a value on the screen. You can print a variable (Python looks up the value it refers to) or a literal value directly.

```python
print(a)    # prints the value that a refers to -> 3
print(123)  # prints a literal value directly
```

```python
age = 25             # int: a whole number
name = "Krish naik"  # str: text wrapped in quotes
```

In Jupyter, the value of the **last expression** in a cell is displayed automatically, even without `print()`.

```python
# No print() needed: Jupyter shows the result of the last line in the cell
age + 10  # age is 25, so this shows 35
```

### 1.1 Naming rules and conventions

**Rules** (breaking these causes a `SyntaxError`):

- A name can contain only letters, digits, and underscores (`_`).
- A name **cannot start with a digit**: `name_1` is valid, `1_name` is not.
- Names are **case-sensitive**: `age`, `Age`, and `AGE` are three different variables.
- Reserved keywords such as `if`, `for`, `class`, and `True` cannot be used as names.

**Conventions** (not enforced by Python, but followed by the community, see [PEP 8](https://peps.python.org/pep-0008/)):

- Use **meaningful** names: `age` tells the reader more than `x`.
- Use **snake_case** for variables: lowercase words joined by underscores, like `first_name`.
- A leading underscore (`_name`) signals that a value is meant for internal use.

```python
# Valid names: letters, digits, and underscores, not starting with a digit
age = 35
name = "monal"

# Invalid names (uncomment a line to see the SyntaxError):
# 1_name = "monal"   # starts with a digit
# first-name = "x"   # hyphens are not allowed
# class = "FDE"      # 'class' is a reserved keyword
```

```python
# Digits are allowed anywhere except as the first character
name_0 = "monal"
name_1 = "pratap"
name_2 = "josephin"

print(name_0)
print(name_1)
print(name_2)
```

```python
# camelCase works, but it is not the Python convention
firstName = "srinu"

# snake_case is the standard style for Python variable names
first_name = "srinu"
```

## 2. Data Types

Every value in Python has a **type**, and the type decides what you can do with the value. Use the built-in function `type(value)` to check it.

| Category | Type | Example |
|---|---|---|
| Numeric | `int` | `12` |
| Numeric | `float` | `12.5` |
| Numeric | `complex` | `13 + 2j` |
| Boolean | `bool` | `True`, `False` |
| Text | `str` | `"hello"` |
| No value | `NoneType` | `None` |

### 2.1 Numbers

- `int`: whole numbers of any size, such as `12`, `-7`, or `1_000_000` (underscores are allowed for readability).
- `float`: numbers with a decimal point, such as `12.0` or `3.14`.
- `complex`: numbers with a real and an imaginary part, written as `a + bj`.

```python
# int: a whole number (positive, negative, or zero)
num_int_1 = 12
print(num_int_1)
print(type(num_int_1))  # <class 'int'>
```

```python
# float: a number with a decimal point
num_float_1 = 12.00083423
print(num_float_1)
print(type(num_float_1))  # <class 'float'>
```

```python
# complex: written as a + bj, where j is the imaginary unit
num_complex_1 = 13 + 2j
print(num_complex_1)
print(num_complex_1.real, num_complex_1.imag)  # real part, imaginary part
print(type(num_complex_1))  # <class 'complex'>
```

### 2.2 Boolean

A `bool` has exactly two possible values: `True` and `False` (note the capital first letter). Comparisons produce booleans, and `if` statements use them to make decisions.

Under the hood, `bool` is a subtype of `int`: `True` behaves like `1` and `False` like `0`.

```python
# bool: only True or False
is_active = True  # try False
print(is_active)
print(type(is_active))  # <class 'bool'>

# True acts like 1 and False acts like 0 in arithmetic
print(True + True)  # 2
```

### 2.3 String

A `str` is a sequence of characters wrapped in single (`'...'`) or double (`"..."`) quotes. Anything inside quotes is a string, even if it looks like a number or a boolean.

```python
# str: text wrapped in single or double quotes
text_1 = "hello"
print(text_1, type(text_1))

# Looks like a bool, but the quotes make it a string
text_2 = 'True'
print(text_2, type(text_2))

# Looks like a number, but the quotes make it a string
text_3 = '123'
print(text_3, type(text_3))

# Spaces inside the quotes are kept exactly as written
text_4 = '123       100'
print(text_4, type(text_4))

# With strings, + joins the text together; it does not add numbers
print(text_3 + text_3)  # '123123', not 246
```

### 2.4 None

`None` means "no value". It is often used as a placeholder before the real value is known, and it is what a function returns when it returns nothing. It is the only value of the type `NoneType`.

```python
# None: a placeholder meaning "no value yet"
value_none_type = None
print(value_none_type)
print(type(value_none_type))  # <class 'NoneType'>
```

### 2.5 Mutable vs immutable data types

Every object has an identity. `id(obj)` returns a number that is unique to that object while it exists (in CPython, it is the object's memory address).

- **Immutable** objects cannot be changed after they are created. When you "change" the variable, Python creates a **new** object and points the name at it, so `id()` changes. Examples: `int`, `float`, `bool`, `str`, `tuple`.
- **Mutable** objects can be changed **in place**, so `id()` stays the same. Examples: `list`, `set`, `dict`.

> The exact `id()` numbers are different every time you run the notebook. What matters is whether two ids are the same or different.

```python
# Immutable example: integers
a = 10
print(id(a))  # identity of the object 10

b = 100
print(id(b))  # a different object, so a different id
```

```python
id_before = id(a)

# An int cannot change, so a + 1 creates a NEW object (11) and points a at it
a = a + 1
print(a)                   # 11
print(id(a))               # a new id
print(id(a) == id_before)  # False: a now refers to a different object
```

```python
# Mutable example: lists
a_list = []  # an empty list
print(id(a_list))
```

```python
id_before = id(a_list)

# append() changes the existing list in place; no new object is created
a_list.append(100)
print(a_list)                   # [100]
print(id(a_list))               # same id as before
print(id(a_list) == id_before)  # True: still the same object
```

## 3. Operators

An **operator** is a symbol that performs an operation on one or more values, called **operands**. In `2 + 3`, the `+` is the operator and `2` and `3` are the operands.

### 3.1 Arithmetic operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | Addition | `10 + 3` | `13` |
| `-` | Subtraction | `10 - 3` | `7` |
| `*` | Multiplication | `10 * 3` | `30` |
| `/` | Division (always returns a `float`) | `10 / 3` | `3.333...` |
| `//` | Floor division (divides, then rounds down) | `10 // 3` | `3` |
| `%` | Modulus (remainder after division) | `10 % 3` | `1` |
| `**` | Exponent (power) | `2 ** 3` | `8` |

```python
value_1 = 10
value_2 = 200

print(value_1 + value_2)  # addition       -> 210
print(value_1 - value_2)  # subtraction    -> -190
print(value_1 * value_2)  # multiplication -> 2000
print(value_2 / value_1)  # division       -> 20.0 (always a float)
```

```python
value_3 = 2
value_4 = 3

print(value_3 % value_4)   # modulus: remainder of 2 / 3        -> 2
print(value_3 ** value_4)  # exponent: 2 to the power of 3      -> 8
print(value_3 // value_4)  # floor division: 2 / 3 rounded down -> 0
```

### 3.2 Comparison operators

Comparison operators compare two values and always return a `bool` (`True` or `False`).

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `2 == 3` | `False` |
| `!=` | Not equal to | `2 != 3` | `True` |
| `>` | Greater than | `2 > 3` | `False` |
| `<` | Less than | `2 < 3` | `True` |
| `>=` | Greater than or equal to | `2 >= 3` | `False` |
| `<=` | Less than or equal to | `2 <= 3` | `True` |

> Don't confuse `=` and `==`: a single `=` **assigns** a value, a double `==` **compares** two values.

```python
value_3 = 2
value_4 = 3

print(value_3 > value_4)   # is 2 greater than 3?             -> False
print(value_3 < value_4)   # is 2 less than 3?                -> True
print(value_3 == value_4)  # is 2 equal to 3?                 -> False
print(value_3 != value_4)  # is 2 not equal to 3?             -> True
print(value_3 >= value_4)  # is 2 greater than or equal to 3? -> False
print(value_3 <= value_4)  # is 2 less than or equal to 3?    -> True
```

### 3.3 Logical operators

Logical operators combine or invert boolean conditions.

| Operator | Result is `True` when... | Example | Result |
|---|---|---|---|
| `and` | both conditions are `True` | `True and False` | `False` |
| `or` | at least one condition is `True` | `True or False` | `True` |
| `not` | the condition is `False` (it flips the value) | `not True` | `False` |

```python
# Problem: a guest can enter a night party only if they are older than 18 AND have an ID

age = 16  # try changing this to 25
has_id = True

can_enter = age > 18 and has_id
# (16 > 18) and True  ->  False and True  ->  False
print(can_enter)
```

```python
# or: True if at least one side is True
is_vip = True
print(can_enter or is_vip)  # False or True -> True
```

```python
# not: flips a boolean value
has_id = not has_id  # not True -> False
print(has_id)
```

### 3.4 Identity operators

Identity operators check whether two names refer to the **same object** in memory (the same `id()`). This is different from `==`, which checks whether two values are **equal**.

| Operator | Result is `True` when... |
|---|---|
| `is` | both names refer to the same object |
| `is not` | the names refer to different objects |

> **Why is `a is c` `True` below?** CPython creates the integers from -5 to 256 once and reuses them, so every `10` in the program is the same object. That is an implementation detail. Use `==` to compare values, and keep `is` for identity checks such as `x is None`.

```python
a = 10
b = 20
c = 10

print(a is b)  # different objects -> False
print(a is c)  # both refer to the same cached object 10 -> True
```

```python
# The same id means the same object
print(id(a))
print(id(b))
print(id(c))  # same as id(a)
```

```python
print(a is not b)  # a and b are different objects -> True
```

```python
# Equal values are not always the same object
list_a = [1, 2, 3]
list_b = [1, 2, 3]

print(list_a == list_b)  # the values are equal -> True
print(list_a is list_b)  # but they are two separate list objects -> False
```

### 3.5 Membership operators

Membership operators check whether a value is present inside a sequence or collection, such as a string, list, tuple, or set.

| Operator | Result is `True` when... |
|---|---|
| `in` | the value is found in the collection |
| `not in` | the value is not found in the collection |

```python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)      # "apple" is in the list     -> True
print("grape" in fruits)      # "grape" is not in the list -> False
print("grape" not in fruits)  # -> True

# With strings, in checks for a substring
print("py" in "python")  # -> True
```

### 3.6 Bitwise operators

Bitwise operators work on the **binary** (base-2) form of integers, one bit at a time.

| Operator | Name | Result |
|---|---|---|
| `&` | AND | bit is 1 only if both bits are 1 |
| `\|` | OR | bit is 1 if either bit is 1 |
| `^` | XOR | bit is 1 if the two bits are different |
| `~` | NOT | flips every bit (`~x` equals `-x - 1`) |
| `<<` | Left shift | moves bits left; each shift multiplies by 2 |
| `>>` | Right shift | moves bits right; each shift floor-divides by 2 |

```python
a = 12  # binary 1100
b = 4   # binary 0100

# AND keeps only the bits that are 1 in both numbers:
#   1100   (12)
# & 0100   (4)
# = 0100   (4)
print(a & b)                       # 4
print(bin(a), bin(b), bin(a & b))  # bin() shows the binary form (0b prefix)

print(a | b)  # OR:  1100 | 0100 = 1100 -> 12
print(a ^ b)  # XOR: 1100 ^ 0100 = 1000 -> 8
```

```python
a = 5  # binary 101

print(a << 1)  # left shift:  101 -> 1010 = 10 (same as 5 * 2)
print(a >> 1)  # right shift: 101 -> 10   = 2  (same as 5 // 2)
```

### 3.7 Operator precedence

When an expression contains several operators, Python evaluates the ones with **higher precedence** first, much like BODMAS in maths. Operators with the same precedence are evaluated left to right, except `**`, which goes right to left. Parentheses always come first, so use them to make your intent clear.

From highest to lowest precedence:

| Order | Operators | Description |
|---|---|---|
| 1 | `()` | Parentheses (grouping) |
| 2 | `**` | Exponent |
| 3 | `+x`, `-x`, `~x` | Unary plus, unary minus, bitwise NOT |
| 4 | `*`, `/`, `//`, `%` | Multiplication, division, floor division, modulus |
| 5 | `+`, `-` | Addition, subtraction |
| 6 | `<<`, `>>` | Bit shifts |
| 7 | `&` | Bitwise AND |
| 8 | `^` | Bitwise XOR |
| 9 | `\|` | Bitwise OR |
| 10 | `==`, `!=`, `>`, `<`, `>=`, `<=`, `is`, `is not`, `in`, `not in` | Comparison, identity, membership |
| 11 | `not` | Logical NOT |
| 12 | `and` | Logical AND |
| 13 | `or` | Logical OR |

Assignment (`=`) happens last: Python evaluates the entire right-hand side, then stores the result in the variable.

```python
# * has higher precedence than +, so the multiplication happens first
result = 2 + 3 * 4
# step 1: 3 * 4  -> 12
# step 2: 2 + 12 -> 14
print(result)
```

```python
# Parentheses are evaluated first
result = (2 + 3) * 4
# step 1: (2 + 3) -> 5
# step 2: 5 * 4   -> 20
print(result)
```

```python
# ** groups right to left: 2 ** 3 ** 2 means 2 ** (3 ** 2) = 2 ** 9
print(2 ** 3 ** 2)  # 512

# ** binds tighter than unary minus: -2 ** 2 means -(2 ** 2)
print(-2 ** 2)  # -4

# Arithmetic first, then comparisons, then not, then and
print(not 2 + 3 > 4 and 10 % 3 == 1)
# step 1: 2 + 3 -> 5          and  10 % 3 -> 1
# step 2: 5 > 4 -> True       and  1 == 1 -> True
# step 3: not True -> False
# step 4: False and True -> False
```

## 4. Collections

A **collection** stores multiple values in a single variable. This section covers the three most common ones.

| Feature | List `[ ]` | Tuple `( )` | Set `{ }` |
|---|---|---|---|
| Ordered (keeps insertion order) | Yes | Yes | No |
| Mutable (can be changed) | Yes | No | Yes (add / remove items) |
| Allows duplicates | Yes | Yes | No |
| Supports indexing (`x[0]`) | Yes | Yes | No |

### 4.1 Lists

A **list** is an ordered, mutable collection. It allows duplicate values and can mix different types.

**Syntax:** `variable = [value_1, value_2, ...]`

Each item has a position called an **index**, starting at `0`. Negative indexes count from the end, so `my_list[-1]` is the last item.

| Method | What it does |
|---|---|
| `append(x)` | Adds `x` to the end of the list |
| `remove(x)` | Removes the first item equal to `x` (error if `x` is missing) |
| `pop(i)` | Removes the item at index `i` and returns it (the last item if `i` is omitted) |
| `index(x)` | Returns the index of the first item equal to `x` |
| `count(x)` | Returns how many times `x` appears |
| `sort()` | Sorts the list in place (all items must be comparable) |
| `reverse()` | Reverses the current order in place |
| `len(my_list)` | Built-in function that returns the number of items |

```python
# A list can hold duplicate values
numbers = [10, 10, 20, 20, 130]
print(numbers)
print(type(numbers))  # <class 'list'>
```

Lists are mutable, so methods like `append()` and `remove()` change the list itself instead of creating a new one.

```python
# append() adds an item to the end of the list
numbers.append(1000)
print(numbers)
```

```python
# Duplicates are allowed: append 1000 a second time
numbers.append(1000)
print(numbers)
```

```python
# remove() deletes an item by its value
numbers.remove(130)
print(numbers)
```

```python
# remove() deletes only the FIRST matching value, so one 1000 is left
numbers.remove(1000)
print(numbers)
```

```python
# Retrieve an item by its index (indexes start at 0)
print(numbers[0])   # first item  -> 10
print(numbers[4])   # fifth item  -> 1000
print(numbers[-1])  # last item   -> 1000
```

```python
# pop(i) removes the item at index i and returns it
removed = numbers.pop(4)
print(removed)  # 1000
print(numbers)
```

```python
numbers.append(45)
print(numbers)

# index(x) returns the position of the first item equal to x
print(numbers.index(45))  # 4
```

A list can mix values of different types. Be careful when searching for booleans: because `True == 1` and `False == 0`, `index(True)` stops at the first `1` it finds.

```python
combination_of_values = [1, 2, 3, 45, "names", "akshay", 78, 0, True]

# True == 1, so index() matches the 1 at position 0, not the True at the end
print(combination_of_values.index(True))   # 0

# False == 0, so index() matches the 0 at position 7
print(combination_of_values.index(False))  # 7
```

```python
# count(x) returns how many times x appears in the list
print(numbers)
print(numbers.count(10))  # 2
```

`sort()` compares items with `<`. Python cannot compare an `int` with a `str`, so sorting a list that mixes them raises a `TypeError`.

```python
dummy_list = ["a", 1, 4, 5, 2, 3, 8, 9, 7, "d", "c"]

# Mixed int and str values cannot be compared, so sort() fails
try:
    dummy_list.sort()
except TypeError as error:
    print("TypeError:", error)
```

```python
# sort() works when all the items can be compared with each other
numbers_only = [1, 4, 5, 2, 3, 8, 9, 7]
numbers_only.sort()  # sorts in place (it returns None, not a new list)
print(numbers_only)
```

```python
# Strings are sorted alphabetically
dummy_list_2 = ['z', 'a', 'c']
dummy_list_2.sort()
print(dummy_list_2)  # ['a', 'c', 'z']
```

```python
# reverse() flips the current order in place (it does not sort)
dummy_list_2.reverse()
print(dummy_list_2)  # ['z', 'c', 'a']
```

```python
# len() returns the number of items in the list
print(len(dummy_list_2))  # 3
```

Because lists are mutable, you can replace an item by its index. The list keeps the same `id()`, which shows it was changed in place.

```python
dummy_list_3 = [1, 2, 3, 4, 5, 6]
print(id(dummy_list_3))
print(dummy_list_3)
```

```python
# Replace the item at index 3 (the fourth item)
dummy_list_3[3] = 100

print(dummy_list_3)      # [1, 2, 3, 100, 5, 6]
print(id(dummy_list_3))  # same id as before: the list was changed in place
```

### 4.2 Tuples

A **tuple** is an ordered, **immutable** collection. Once it is created, items cannot be added, removed, or replaced. Use a tuple for data that should not change.

**Syntax:** `variable = (value_1, value_2, ...)`

Tuples support indexing just like lists, and they have only two methods: `index(x)` and `count(x)`.

```python
tuple_1 = (1, 2, 3, 4, 5, 6)

# Indexing works the same way as with lists
print(tuple_1[2])  # 3
```

```python
# Tuples are immutable, so item assignment raises a TypeError
try:
    tuple_1[2] = 100
except TypeError as error:
    print("TypeError:", error)
```

Parentheses alone do not make a tuple; the **comma** does. A tuple with a single item needs a trailing comma.

```python
# Without a comma the parentheses only group the value, so this is a str
not_a_tuple_1 = ("MONDAY")
print(type(not_a_tuple_1))  # <class 'str'>

# The trailing comma makes it a tuple with one item
this_is_tuple_1 = ("MONDAY",)
print(type(this_is_tuple_1))  # <class 'tuple'>
```

```python
print(tuple_1.index(6))  # position of the first 6 -> 5
print(tuple_1.count(6))  # how many times 6 appears -> 1
```

**List or tuple?** Choose based on whether the data is expected to change.

```python
# FDE classes are always on the weekend, so the days never change: use a tuple
fde_class_days = ("Saturday", "Sunday")

# New topics are added every week, so the data changes: use a list
topics = ["variables", "data types"]
topics.append("operators")

print(fde_class_days)
print(topics)
```

### 4.3 Sets

A **set** is an unordered collection of **unique** items. Duplicate values are dropped automatically. You can add and remove items, but each item must itself be immutable (for example a number, string, or tuple).

**Syntax:** `variable = {value_1, value_2, ...}`

- Use `set()` to create an empty set. Empty braces `{}` create an empty **dictionary**, not a set.
- Sets have no order, so you cannot access items by index, and printing a set may show the items in any order.

| Method | What it does |
|---|---|
| `add(x)` | Adds `x` (does nothing if `x` is already in the set) |
| `remove(x)` | Removes `x` (raises a `KeyError` if `x` is missing) |
| `discard(x)` | Removes `x` (does nothing if `x` is missing) |

```python
# Duplicate values are removed automatically
numbers = {1, 1, 2, 3, 4, 5, 6, 6, 7, 8}
print(numbers)
print(type(numbers))  # <class 'set'>
```

```python
# set() creates an empty set
numbers_set_1 = set()
print(numbers_set_1)
print(type(numbers_set_1))  # <class 'set'>

# Empty braces create a dictionary, not a set
print(type({}))  # <class 'dict'>
```

```python
# add() inserts one item
numbers_set_1.add(4)
print(numbers_set_1)
```

```python
numbers_set_1.add(400)
print(numbers_set_1)

# 4 is already in the set, so adding it again changes nothing
numbers_set_1.add(4)
print(numbers_set_1)
```

```python
# A set can mix types; the printed order is not guaranteed
numbers_set_1.add("ajay")
print(numbers_set_1)
```

```python
# Sets are unordered, so there is no index to look up
try:
    numbers_set_1[1]
except TypeError as error:
    print("TypeError:", error)
```

```python
# discard() removes an item by its value
numbers_set_1.discard(400)
print(numbers_set_1)

# discard() does nothing if the item is missing (remove() would raise a KeyError)
numbers_set_1.discard(999)
print(numbers_set_1)
```

## Summary

- A **variable** is a name that refers to a value. Python works out the type from the value.
- Names can't start with a digit or be a keyword. Use meaningful **snake_case** names.
- The core data types are `int`, `float`, `complex`, `bool`, `str`, and `NoneType`. Check a value's type with `type()`.
- **Immutable** objects (`int`, `str`, `tuple`) are replaced by a new object when "changed". **Mutable** objects (`list`, `set`) are changed in place.
- Use `==` to compare **values** and `is` to compare **identity** (mainly `x is None`).
- Precedence, from highest to lowest: `()` → `**` → `* / // %` → `+ -` → comparisons → `not` → `and` → `or`.
- **List**: ordered and mutable. **Tuple**: ordered and immutable. **Set**: unordered, with unique items.
